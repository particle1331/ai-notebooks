# Agentic Systems

An **AI agent** is an LLM equipped with tools, memory, and a control loop that lets it pursue goals across multiple steps &mdash; perceiving its environment, deciding what to do next, acting, and updating based on the result. This series builds that picture from first principles: starting with the foundational papers that defined the field, through the mechanisms that make agents work (tool calling, reasoning, context management), up to the patterns used to compose them into reliable multi-step systems.

The emphasis is on [conceptual depth]{.underline} over application breadth. Each notebook is organized around a specific "how does X work?" question, answered with code experiments that produce observable, measurable outcomes.

## About This Series

This series is a technical deep-dive into the concepts underlying agentic AI systems &mdash; not a tutorial on building applications. Every section has a runnable experiment that makes the concept concrete and measurable. The goal is to build intuition that transfers to new problems, not familiarity with a specific framework.

**Audience:** Someone who has built or closely read an LLM-powered application and wants to understand *why* the design choices work, when they break down, and what the alternatives are. Familiarity with the OpenAI Python SDK and basic async Python is assumed.

The series proceeds from mechanisms to patterns: first establishing how individual components work (reasoning, tools, context, memory), then examining how those components combine into higher-order agentic patterns.

## What You'll Build Understanding Of

By the end of this series, you will understand:

- Why chain-of-thought prompting improves multi-step reasoning, measured empirically on GSM8K
- How the ReAct loop works at the message level, and why interleaving reasoning and tool calls reduces hallucination
- How Toolformer trained a model to call APIs via self-supervised data generation &mdash; the origin of modern tool-calling
- How the Generative Agents architecture combines memory, reflection, and planning for believable long-horizon behavior
- How the OpenAI function-calling API works end-to-end, including manual tool dispatch for providers without native support
- How to design memory systems that balance recency, importance, and relevance for retrieval
- How the evaluator-optimizer (reflection) pattern improves output quality iteratively, and when it saturates

## How to Read This Series

**If you're starting from scratch:** Read 00 (Foundations) first for conceptual grounding, then 01 and 02 for the API mechanics. Everything else builds on these three.

**If you already understand tool calling:** Jump to 06 (Memory) or the Patterns subfolder. Each notebook is designed to stand on its own with minimal back-references.

**If you want practical patterns:** Start with [Patterns &mdash; Reflection](patterns/01-reflection.ipynb) &mdash; it is the most immediately applicable notebook and uses the infrastructure built in 01 and 02.

**Next steps:** The concepts here feed directly into the more advanced topics planned for this series: reasoning patterns (ReAct in depth, Tree of Thoughts), context engineering, model selection and routing, reasoning models (o1/R1), and multi-agent orchestration.
