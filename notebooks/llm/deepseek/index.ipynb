# DeepSeek

DeepSeek-V2 and V3 introduced two architectural ideas that matter at scale: Multi-head Latent Attention (MLA), which compresses the KV cache by projecting keys and values through a low-rank bottleneck, and Mixture-of-Experts (MoE), which replaces the dense FFN with a sparse routing mechanism that activates only a fraction of parameters per token. The Engram layer adds a third axis — hash-addressed n-gram memory — that augments the residual stream without attention overhead. These notebooks implement all three from scratch.

## About This Series

**Audience.** Someone who has completed [NB01 (GPT Architecture)](/notebooks/llm/01-gpt-architecture.html) from the main series. The notebooks modify and extend the baseline GPT, so familiarity with causal self-attention, RMSNorm, and the GPT block is assumed.

**Stack.** Pure PyTorch. NanoDeepSeek trains on TinyShakespeare or FineWeb-Edu-10B at ~40M activated parameters. Honest caveats are given wherever the benefits of MLA, MoE, or Engram only manifest at production scale (10B+ parameters).

**Scope.** These notebooks are [architecture and mechanics notebooks]{.mark}, not empirical scaling studies. The goal is to understand how each component works and how to implement it correctly — not to reproduce DeepSeek's benchmark numbers.
