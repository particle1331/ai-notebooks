# Language Models From Scratch

**Large language models** represent the most significant advance in AI since deep learning itself. GPT-3, PaLM, LLaMA, and DeepSeek are not magic — they are Transformers trained with well-understood objectives on well-understood data pipelines, aligned with well-understood techniques. This series makes all of that explicit.

<br>

## About This Series

This series teaches you how to build, train, debug, and align language models from first principles — in pure PyTorch, with no framework magic hiding the mechanics.

**Audience:** Someone who has completed the [Deep Learning Foundations](../deep/index.ipynb) series (or equivalent) and wants to go from "I understand backpropagation" to "I can train and align a language model from scratch." A mathematics background is assumed — derivations are shown in full, not waved away. Code is the other half of the explanation.

Every concept is grounded in code you write yourself. The tutorials describe the Transformer forward pass in full, show how to build a pretraining data pipeline, and implement GRPO from scratch. The goal is to ensure you understand every moving part well enough to debug a broken training run at 3am.

## What You'll Build Understanding Of

By the end of this series, you will understand:

- How the Transformer architecture works from the matrix multiplications up — attention, RoPE, causal masking, FFN, layer norm placement, and why each choice exists
- How to instrument a training run so you can *see* what's happening — gradient distributions, weight norms, activation statistics, loss spikes — in real time
- How to diagnose and fix the pathologies that kill training runs: vanishing gradients, dead neurons, loss divergence, thrashing LR schedules
- How pretraining works at the data pipeline level — packing, shuffling, deduplication — and how scaling laws tell you how much compute to spend
- How supervised fine-tuning, LoRA, DPO, and GRPO work mathematically and in code — including the full reward model training pipeline
- How quantization trades precision for memory, what it costs in perplexity, and when each format (BF16, INT8, INT4) is appropriate
- How to build a KV cache, why Flash Attention is faster, and what speculative decoding does
- How to assemble all of the above into a single coherent pipeline that trains a small reasoning model from scratch

## Prerequisites

- Python 3.10+, PyTorch 2.0+, NumPy, Matplotlib
- Linear algebra (matrix products, eigenvalues), calculus (chain rule, partial derivatives), basic probability (expectation, KL divergence). Derivations are shown in full.
- A machine with at least one CUDA GPU is recommended for Part III onward; Parts I and II run fine on CPU.
- Completion of (or familiarity with the content of) the [Deep Learning Foundations](../deep/index.ipynb) series.

## How to Read This Series

**If you want to understand transformers before anything else:** Read Tutorials 2 and 3 carefully. That's the minimum to reason about architecture decisions.

**If you're debugging a training run right now:** Jump to Part II (Tutorials 4–6). The gradient hooks tutorial is the most immediately practical thing in the series.

**If you're designing a pretraining run:** Part III is the core — data pipelines, optimizers, mixed precision, and the scaling law tutorial will tell you how to allocate compute.

**If you want a complete reference:** Every tutorial ends with a summary table covering the key formulas, API surfaces, and decision criteria from that tutorial. They are designed to stand alone as quick references.

**If you're aligning a model:** Read Part IV in order. DPO (12) depends on understanding SFT (11). GRPO (14) depends on understanding reward models (13).
