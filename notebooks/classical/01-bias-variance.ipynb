## Double Descent

The classical U-curve predicts that increasing model complexity beyond the sweet spot raises test error. Modern overparameterized models -- neural networks with millions of parameters trained on thousands of examples -- routinely violate this prediction. See [Double Descent](/notebooks/deep/18-appendix-double-descent.html) in the Deep Learning Foundations appendices for a full treatment, including reproduction of the Nakkiran et al. curves on ResNet / CIFAR-10.
