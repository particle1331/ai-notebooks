# AI Engineering Crash Course

This series is a structured preparation path for [mid-career engineers]{.mark} looking to move into [senior AI engineering roles]{.mark}[^clarify] — particularly in regulated domains such as financial services, where LLM integration intersects with compliance obligations, audit requirements, and strict reliability targets. The assumption is that you arrive with solid software engineering fundamentals and ML experience but want to close the gap between what you know and what production & enterprise AI engineering actually demands.

The series is organized into two paths. The **Crash Course** (notebooks 01–13) is a linear progression from API fundamentals through evaluation, agents, observability, fine-tuning, model serving, cost optimization, and infrastructure — culminating in a **capstone project**: a fully deployed, publicly accessible compliance document reviewer that synthesizes every skill in the series. The **[Deep Dives](/notebooks/prep/deep-dives/index.html)** are standalone notebooks that extend specific topics from the main path for readers who want to go further on a particular subject.

[^clarify]: There is no standard definition of a "senior engineer". In some context, this can mean Sr. engineer to Staff or Principal. In any case, the series will prepare the reader for advanced roles that demand technical expertise across the full LLM system stack — from API integration and evaluation through infrastructure, enterprise security, and production deployment.

## Learning Paths

Two tracks, one dependency graph:

| Track | Notebooks | Commitment | Outcome |
|---|---|---|---|
| **Crash Course** | 01–13 | 1 week (dense, 3–4 hr/day) | End-to-end deployed AI system |
| **Extended Main Path** | 01–13 (paced) | 1 month | Same outcome, deeper absorption |
| **Deep Dives** | DD:01–08 (any order) | +1–3 months | Expertise in chosen sub-domains |

The crash course covers one notebook per day. Notebooks 01–03 are review on Day 1; the new material runs Days 2–7, then the extended path picks up weeks 2–4 at a more deliberate pace.

## Capstone: Compliance Document Reviewer {#sec-capstone}

[Notebook 13](/notebooks/prep/13-capstone-compliance.html) is the series capstone. It builds and deploys a production-grade LLM-powered compliance document reviewer targeting financial services use cases. The system accepts a contract, regulatory filing, or disclosure document and returns a structured compliance report: flagged issues with severity scores, citations to relevant regulatory passages, and recommended remediation actions — all with a complete audit trail.

Every notebook in the series contributes a component:

| Capstone Component | Draws From |
|---|---|
| PII detection and redaction (pre-LLM) | [02](/notebooks/prep/02-prompt-engg.html) (security mindset), [DD:06](/notebooks/prep/deep-dives/06-ai-security.html) (AI security) |
| Model routing: triage tier vs. review tier | [01](/notebooks/prep/01-llm-api.html) (API), [11](/notebooks/prep/11-cost-optimization.html) (cost optimization) |
| Policy RAG store (SEC/FINRA rules) | [03](/notebooks/prep/03-rag-concepts.html), [07](/notebooks/prep/07-rag-pipeline.html) (ChromaDB + hybrid retrieval) |
| LangGraph review pipeline | [08](/notebooks/prep/08-agents-langgraph.html) (multi-agent) |
| Structured compliance report (Pydantic) | [01](/notebooks/prep/01-llm-api.html) (structured output) |
| Langfuse audit trace (cost, latency, doc hash) | [06](/notebooks/prep/06-observability.html) (observability) |
| Automated eval regression suite | [04](/notebooks/prep/04-eval-concepts.html) (eval concepts), [DD:02](/notebooks/prep/deep-dives/02-eval-pipelines.html) (eval pipelines) |
| Dockerized FastAPI service | [10](/notebooks/prep/10-model-serving.html) (serving), [12](/notebooks/prep/12-ai-infra.html) (infra) |

| Secrets Manager (API keys, no hardcoded secrets) | [12](/notebooks/prep/12-ai-infra.html) (security) |

The finished artifact is a publicly accessible HTTPS endpoint you can put in a portfolio or demonstrate in a system design interview.

## Prerequisites

- **Python:** Comfortable with classes, decorators, async/await, type hints, and Pydantic models.
- **ML fundamentals:** Embeddings, transformers at a conceptual level, supervised learning.
- **APIs:** REST API consumption, JSON handling, environment variables for secrets.
- **Git:** Basic branching and commit workflow.

The series does not assume prior LLM engineering experience. It does assume you learn by reading and running code — every notebook is executable top to bottom.

:::{.callout-note}
An [OpenAI API key]{.underline} is required from notebook 01 onwards. Notebooks 12 and 13 additionally require an AWS account with permissions to create ECS, ECR, VPC, IAM, and ACM resources. Estimated AWS cost for the capstone deployment is under $5/month on Fargate free tier.
:::

## How to Use This Series

**If you have one week:** Follow the crash course schedule above. Run every cell. Do not skip the exercises at the end of each notebook — they are short implementation tasks designed to surface gaps before an interview.

**If you have one month:** Follow the same order at a slower pace. After each notebook, read the corresponding [deep dive](/notebooks/prep/deep-dives/index.html) if the topic is one of your target areas. Use the [capstone](/notebooks/prep/13-capstone-compliance.html) as a week-4 project.

**If you are preparing for a system design interview:** Read the capstone architecture section ([13](/notebooks/prep/13-capstone-compliance.html)) first to understand where the series is heading, then work forward from [notebook 01](/notebooks/prep/01-llm-api.html). Every design decision in the capstone is motivated by a concept introduced in an earlier notebook.

**If you are already strong in one area:** The series is modular from [notebook 04](/notebooks/prep/04-eval-concepts.html) onwards. If you are solid on RAG fundamentals ([03](/notebooks/prep/03-rag-concepts.html)), skip ahead to evaluation ([04](/notebooks/prep/04-eval-concepts.html)) or agents ([05](/notebooks/prep/05-agents-foundations.html)). Cross-references in each notebook tell you what to review if something is unclear.
