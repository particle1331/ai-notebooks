## About This Series

This series builds the mathematical and computational foundations of deep learning from scratch — in pure Python and PyTorch, deriving everything from first principles.

**Audience:** Someone comfortable with Python, NumPy, and basic calculus who wants to understand how neural networks actually work before moving on to language models. A mathematics background is assumed — derivations are shown in full, not waved away. Code is the other half of the explanation.

Every concept is grounded in code you write yourself. The series covers the full story from a single linear classifier through multi-layer networks, then builds up to automatic differentiation and modern gradient-based optimizers. This is the prerequisite foundation for the [Language Models from Scratch](../llm/index.ipynb) series.

## What You'll Build Understanding Of

By the end of this series, you will understand:

- How softmax regression and cross-entropy loss work from first principles, including gradient derivations
- How multi-layer perceptrons and backpropagation work, with every matrix dimension accounted for
- How computational graphs enable automatic differentiation, and how PyTorch's autograd engine implements reverse-mode AD
- How to build training loops in PyTorch — tensors, gradients, parameter updates
- How gradient-based optimizers work — from vanilla gradient descent through Momentum, RMSProp, Adam, and AdamW — and how to tune learning rate schedules and batch size

## Prerequisites

- Python 3.10+, PyTorch 2.0+, NumPy, Matplotlib
- Linear algebra (matrix products, eigenvalues), calculus (chain rule, partial derivatives). Derivations are shown in full.
- Everything in this series runs on CPU — no GPU required.

## How to Read This Series

**If you're new to deep learning:** Work through 01 → 02 → 03 in order. They build on each other. Then do Tutorial 01 and 01b for the PyTorch-native view.

**If you already understand backpropagation:** Tutorial 01b (optimizer mechanics) stands alone and is the most practically useful reference in the series.

**Next steps:** After completing this series, continue with [Language Models from Scratch](../llm/index.ipynb).
