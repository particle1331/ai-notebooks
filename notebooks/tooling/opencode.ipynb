# OpenCode: AI Coding Agent in the Terminal

Learning to use [OpenCode](https://opencode.ai), an open source AI coding agent available as a terminal-based interface, desktop app, or IDE extension.

## Introduction

OpenCode is a powerful AI-powered coding assistant designed for the terminal. Unlike traditional IDE extensions, it operates as a full-featured TUI (Terminal User Interface) that allows you to ask questions about your codebase, request features, make changes, and iterate on implementations.

Key benefits:
- Native terminal experience with no IDE dependencies
- Direct access to your entire codebase context
- Support for multiple LLM providers (OpenAI, Anthropic, Groq, etc.)
- Git integration for tracking changes
- Shareable conversations with your team
- Customizable commands, tools, and keybinds

## Installation & Setup

### Install OpenCode

Using the install script:
```bash
curl -fsSL https://opencode.ai/install | bash
```

Or via package managers:
```bash
# macOS/Linux
brew install anomalyco/tap/opencode

# Node.js
npm install -g opencode-ai

# Arch Linux
sudo pacman -S opencode
```

### Configure LLM Provider

OpenCode supports multiple LLM providers. Recommended: Use [OpenCode Zen](https://opencode.ai/docs/zen) for curated models.

1. Run `/connect` in the TUI
2. Select provider and authenticate
3. Add billing details and copy API key
4. Paste API key into OpenCode

### Initialize for Your Project

```bash
cd /path/to/project
opencode
/init
```

This creates an `AGENTS.md` file that helps OpenCode understand your project structure and coding patterns.

## Basic Usage

### Modes

OpenCode has two primary modes, toggled with **Tab**:

**Plan Mode** (read-only)
- OpenCode suggests how to implement features
- No changes are made to your codebase
- Good for reviewing implementation strategies before committing

**Build Mode** (write mode)
- OpenCode makes actual changes to your code
- Creates commits with your changes
- Default mode for active development

### Common Tasks

**Ask Questions**
```
How is authentication handled in @src/api/auth.ts?
```
Use `@` to reference files (fuzzy search enabled).

**Add Features**
1. Switch to Plan Mode (Tab)
2. Describe what you want
3. Review the suggested implementation
4. Switch back to Build Mode (Tab)
5. Ask OpenCode to make the changes

**Make Changes**
```
Refactor the function in @src/utils.ts to use modern async/await patterns.
```

**Undo Changes**
```
/undo
```
Reverts changes and shows original message for iteration.

## Workflow Tips

### File References

Use the `@` symbol to reference files:
- Fuzzy search enabled (partial names work)
- Include full paths for clarity
- Example: `@src/components/Button.tsx`

### Plan Before Building

For complex features:
1. Use **Plan Mode** to get OpenCode's suggested approach
2. Ask clarifying questions
3. Iterate on the plan
4. Switch to **Build Mode** and execute

This ensures better implementations and fewer revisions.

### Context & Clarity

Give OpenCode plenty of context:
- Explain *why* you need a change, not just *what*
- Reference related code or patterns
- Show examples of desired output
- You can drag/drop images for visual reference

## Built-in Commands

| Command   | Description                                               |
|-----------|-----------------------------------------------------------|
| `/init`   | Initialize OpenCode for a project (creates AGENTS.md)    |
| `/undo`   | Undo the last change(s)                                   |
| `/redo`   | Redo a change that was undone                             |
| `/share`  | Create a shareable link to your conversation              |
| `/help`   | Show available commands and help                          |

You can also create custom commands in `.opencode/commands/` directory (see [Commands documentation](https://opencode.ai/docs/commands)).

## Advanced Features: Agents

OpenCode has a sophisticated agent system that allows you to create specialized AI assistants for different tasks.

### Agent Types

**Primary Agents**
- Main assistants you interact with directly
- Switch between them with **Tab** key
- Built-in: **Build** (all tools enabled) and **Plan** (restricted mode for analysis)

**Subagents**
- Specialized assistants invoked for specific tasks
- Can be invoked by primary agents automatically or manually with `@` mention
- Built-in: **General** (research and multi-step tasks) and **Explore** (fast, read-only codebase exploration)

### Built-in Agents

| Agent | Type | Purpose |
|-------|------|----------|
| **Build** | Primary | Default development agent with full tool access |
| **Plan** | Primary | Planning and analysis in read-only mode |
| **General** | Subagent | Research and multi-step task execution |
| **Explore** | Subagent | Fast codebase exploration (read-only) |

### Using Subagents

**Automatically**: Primary agents invoke them when needed for specialized tasks.

**Manually**: Mention a subagent with `@`:
```
@explore find all uses of the deprecated auth function
@general help me refactor this module in multiple steps
```

### Creating Custom Agents

Create a new agent interactively:
```bash
opencode agent create
```

This will:
1. Ask where to save (global or project-specific)
2. Request a description of the agent's purpose
3. Generate an appropriate system prompt
4. Let you configure tool access
5. Create a markdown file in `.opencode/agents/`

### Configuring Agents

**In opencode.json:**
```json
{
  "agent": {
    "code-reviewer": {
      "description": "Reviews code for best practices",
      "mode": "subagent",
      "model": "anthropic/claude-sonnet",
      "temperature": 0.1,
      "permission": {
        "edit": "deny",
        "bash": "allow"
      }
    }
  }
}
```

**In markdown** (`.opencode/agents/review.md`):
```markdown
---
description: Code review without modifications
mode: subagent
temperature: 0.1
permission:
  edit: deny
  bash: allow
---

You are a code reviewer. Focus on quality, performance, and security.
```

### Key Configuration Options

- **description**: What the agent does and when to use it
- **mode**: `primary`, `subagent`, or `all`
- **model**: Override the default LLM model for this agent
- **temperature**: Control response randomness (0.0-1.0)
- **permission**: Control tool access (`ask`, `allow`, `deny`)
- **steps**: Max number of agentic iterations before text-only response

### Use Cases

- **Code Reviewer**: Read-only agent for reviewing PRs and suggesting improvements
- **Documentation Writer**: Agent optimized for writing and maintaining docs
- **Security Auditor**: Specialized agent for security analysis
- **Debug Assistant**: Focused on investigation with bash and read tools
- **Fast Explorer**: Quick codebase searches without heavy processing

## Customization

### Configuration File

OpenCode uses `opencode.json` (or `opencode.jsonc` for comments) in your project root:

```json
{
  "$schema": "https://opencode.ai/config.json",
  "instructions": ["AGENTS.md", ".github/AI_PREFERENCES.md"]
}
```

### AGENTS.md

A critical file that tells OpenCode about your project:
- Build and development commands
- Code style guidelines and naming conventions
- Testing and validation approaches
- Project-specific utilities and structure
- Git workflow preferences
- AI assistant preferences (output handling, cell references, etc.)

Create one with `/init` or write manually.

### Keybinds, Themes, Formatters

Customize with configuration options:
- **Themes**: Choose from built-in themes
- **Keybinds**: Customize keyboard shortcuts
- **Formatters**: Configure code formatters (prettier, black, etc.)
- **Tools**: Define custom tools for your workflow

See [Configuration documentation](https://opencode.ai/docs/config) for details.

## Advanced Workflow: byobu for Multi-Pane Development

For a more organized development experience, use [byobu](https://www.byobu.org/) as a wrapper around tmux. Byobu provides intuitive keybindings and a polished status bar.

### Setup

```bash
# Install byobu
sudo apt install byobu      # Debian/Ubuntu
brew install byobu          # macOS

# Start byobu
byobu
```

### Window Management

| Shortcut | Action                          |
|----------|--------------------------------|
| `F2`     | Create new window              |
| `F3` / `F4` | Switch between windows      |
| `F8`     | Rename current window          |
| `F6`     | Detach from byobu session      |

### Pane Management (within a window)

| Shortcut | Action                         |
|----------|--------------------------------|
| `Ctrl+F2` | Create vertical split          |
| `Shift+F2` | Create horizontal split       |
| `Shift+F3` / `Shift+F4` | Switch between panes |
| `Ctrl+F6` | Kill current pane              |

### Broadcast Commands to All Panes

| Shortcut | Action                         |
|----------|--------------------------------|
| `Shift+F9` | Execute command on all panes in current window |

This opens a prompt where you can type a command that will execute in all panes simultaneously.

### Recommended Workflow for OpenCode

**Step 1: Start byobu**
```bash
byobu
```

**Step 2: Create two windows**
- Press `F2` to create first window (for OpenCode)
- Press `F2` again to create second window (for terminal/testing)

**Step 3: Rename windows**
- In OpenCode window: press `F8`, type `opencode`, press Enter
- Switch to terminal window: press `F4`
- Press `F8`, type `terminal`, press Enter

**Step 4: Set up terminal window with splits**
- Press `Ctrl+F2` to create a vertical split
- Press `Shift+F2` for horizontal splits as needed
- Use `Shift+F3`/`Shift+F4` to switch between panes

**Step 5: Run OpenCode in the first window**
- Press `F3` to switch to opencode window
- Type `opencode` and start working

**Step 6: Work in terminal window as needed**
- Press `F4` to switch to terminal window
- Run tests, git commands, etc.
- Use `Shift+F9` to broadcast commands to all panes if testing across multiple environments

**Step 7: Detach and reattach**
- Press `F6` to detach from byobu
- Type `byobu attach` to return to your session

### Benefits of This Setup

- **Organized**: OpenCode and development tools in separate windows
- **Flexible**: Easy to create splits for multiple testing environments
- **Persistent**: Your layout survives terminal restarts
- **Powerful**: Broadcast commands to all panes for parallel testing
- **Intuitive**: Function key bindings are easier to remember than tmux sequences

## Appendix: Useful Terminal Tools

While learning OpenCode and terminal workflows, you'll benefit from these tools:

### fzf - Fuzzy Finder

**What it is:** A command-line fuzzy finder that enables fast, interactive searching and navigation.

**Install:**
```bash
brew install fzf
```

**Verify installation:**
```bash
which fzf
fzf --version
```

**Usage:**
One of the best uses is fuzzy directory navigation:
```bash
cd **<TAB>
```

This activates fzf's fuzzy completion, allowing you to:
1. Type partial directory names
2. Use arrow keys or fzf search to navigate
3. Press Enter to cd into the selected directory

**Other useful fzf features:**
- History search: `Ctrl+R` (in bash/zsh with fzf integration)
- File selection: `vim **<TAB>` to fuzzy search and open files
- Kill process: `kill -9 **<TAB>` to fuzzy search and kill processes

**Why it's useful:**
- Faster navigation than typing full paths
- Partial matching works great
- Integrates seamlessly with shell completions
- Part of many developer workflows (similar to opening projects in VS Code)
