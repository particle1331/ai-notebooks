# Special Topics & Tooling

On technical talks, practical tools, and emerging topics across data, ML, and engineering.

## Contents

| Topic             | Summary                                                     |
|-------------------|-----------------------------------------------------------------|
| [01 · Runpod Setup](01-runpod.ipynb) | Step-by-step guide to setting up SSH access, GitHub authentication, and Python environments on [Runpod.io](console.runpod.io) **pods**. Useful stuff like persistent terminal sessions using ([tmux](https://tmuxcheatsheet.com/)/[byobu](https://www.byobu.org/)) and setting up [Quatro docs](https://quarto.org/) are also covered. |
| [02 · OpenCode](02-opencode.ipynb) | OpenCode: AI coding agent in the terminal. |
| [03 · Singular Value Decomposition](03-svd.ipynb) | Derivation of the SVD from first principles via eigen-decomposition, manual implementation validated against NumPy, and application to low-rank matrix approximation. |
| [04 · MCP Servers](04-mcp.ipynb) | MCP (Model Context Protocol) servers with FastMCP — in-memory, HTTP remote, and local subprocess modes. |
| [05 · Security Tooling](05-security.ipynb) | Defense-in-depth security for Python projects: pre-commit secrets scanning with Gitleaks, dependency vulnerability checking with pip-audit and Dependabot, CI gates via GitHub Actions, workflow hardening with SHA-pinned Actions and zizmor, static analysis with Bandit, and runtime secret management with python-dotenv. |
| [06 · RL Foundations](06-rl-foundations.ipynb) | MDPs, Bellman equations, dynamic programming, and model-free control (MC, TD, Q-learning) on a from-scratch grid world. Establishes the value function and advantage function that underlie all modern alignment algorithms. |
| [07 · Policy Gradients](07-rl-policy-gradients.ipynb) | The policy gradient theorem, REINFORCE, variance reduction via baselines, and Advantage Actor-Critic (A2C) on CartPole. Each step progressively reduces gradient variance while preserving unbiasedness. |
| [08 · Proximal Policy Optimization](08-rl-ppo.ipynb) | Generalized Advantage Estimation (GAE), trust regions, and PPO from scratch. Full implementation on CartPole with diagnostic plots and ablations. PPO is the direct precursor to RLHF and GRPO. |
| [09 · RL for Language Models](09-rl-for-llms.ipynb) | LLM generation as an MDP, the KL-regularized RLHF objective and its closed-form optimal policy, DPO as implicit RL, PPO-for-LLMs anatomy, GRPO derivation, reward hacking simulation, and a unified view of the alignment landscape. |
