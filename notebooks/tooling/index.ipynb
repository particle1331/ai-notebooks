# Special Topics & Tooling

On technical talks, practical tools, and emerging topics across data, ML, and engineering.

## Contents

| Topic             | Summary                                                     |
|-------------------|-----------------------------------------------------------------|
| [Runpod Setup](01-runpod.ipynb) | Setting up SSH access, GitHub authentication, and Python environments on [Runpod.io](console.runpod.io) pods. Useful stuff like persistent terminal sessions using ([tmux](https://tmuxcheatsheet.com/)/[byobu](https://www.byobu.org/)) and setting up [Quarto docs](https://quarto.org/) are also covered. |
| [Weak Supervision](02-weak-supervision.ipynb) | Data programming with labeling functions: generative model for learning LF parameters via MLE, soft label inference, and noise-aware discriminative training. Applied to synthetic 2D data and IMDB sentiment classification. |
| [Prompt Optimization](03-prompt-optimization.ipynb) | Automatically improve prompts from ground truth labels alone — no rubric required. Covers OPRO (instruction search via LLM proposals), Optuna (few-shot selection via Bayesian optimization), and DSPy (a framework that unifies both searches), with a full comparison on essay grading. |
| [Singular Value Decomposition](04-svd.ipynb) | Derivation of the SVD from first principles via eigen-decomposition, manual implementation validated against NumPy, and application to low-rank matrix approximation. |
| [MCP Servers](05-mcp.ipynb) | MCP (Model Context Protocol) servers with [FastMCP](https://gofastmcp.com/getting-started/welcome) — in-memory, HTTP remote, and local subprocess modes. |
| [Security Tooling](06-security.ipynb) | Layered security for Python projects: secrets scanning, dependency auditing, CI gates, workflow hardening, and static analysis. |
| [Text-to-LoRA](11-text-to-lora.ipynb) | Compile natural language specifications into neural programs that run locally. Covers LoRA from scratch, hypernetworks, the text-to-LoRA training objective, and a full toy pipeline ending in a serializable callable. |

## Reinforcement Learning

A self-contained four-notebook series building Reinforcement Learning (RL) from first principles through modern LLM alignment. Each notebook extends the previous one, culminating in the theoretical foundations of RLHF, PPO-for-LLMs, and GRPO.


| Topic             | Summary                                                     |
|-------------------|-----------------------------------------------------------------|
| [RL1: Foundations](07-rl-foundations.ipynb) | MDPs, Bellman equations, dynamic programming, and model-free control (MC, TD, Q-learning) on a from-scratch grid world. Establishes the value function and advantage function that underlie all modern alignment algorithms. |
| [RL2: Policy Gradients](08-rl-policy-gradients.ipynb) | The policy gradient theorem, REINFORCE, variance reduction via baselines, and Advantage Actor-Critic (A2C) on CartPole. Each step progressively reduces gradient variance while preserving unbiasedness. |
| [RL3: Proximal Policy Optimization](09-rl-ppo.ipynb) | Generalized Advantage Estimation (GAE), trust regions, and PPO from scratch. Full implementation on CartPole with diagnostic plots and ablations. PPO is the direct precursor to RLHF and GRPO. |
| [RL4: RL for Language Models](10-rl-for-llms.ipynb) | LLM generation as an MDP, the KL-regularized RLHF objective and its closed-form optimal policy, DPO as implicit RL, PPO-for-LLMs anatomy, GRPO derivation, reward hacking simulation, and a unified view of the alignment landscape. |
