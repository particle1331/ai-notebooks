## Reinforcement Learning

A self-contained four-notebook series building Reinforcement Learning (RL) from first principles through modern LLM alignment. Each notebook extends the previous one, culminating in the theoretical foundations of RLHF, PPO-for-LLMs, and GRPO.

| Topic             | Summary                                                     |
|-------------------|-----------------------------------------------------------------|
| [RL1: Foundations](06-rl-foundations.ipynb) | MDPs, Bellman equations, dynamic programming, and model-free control (MC, TD, Q-learning) on a from-scratch grid world. Establishes the value function and advantage function that underlie all modern alignment algorithms. |
| [RL2: Policy Gradients](07-rl-policy-gradients.ipynb) | The policy gradient theorem, REINFORCE, variance reduction via baselines, and Advantage Actor-Critic (A2C) on CartPole. Each step progressively reduces gradient variance while preserving unbiasedness. |
| [RL3: Proximal Policy Optimization](08-rl-ppo.ipynb) | Generalized Advantage Estimation (GAE), trust regions, and PPO from scratch. Full implementation on CartPole with diagnostic plots and ablations. PPO is the direct precursor to RLHF and GRPO. |
| [RL4: RL for Language Models](09-rl-for-llms.ipynb) | LLM generation as an MDP, the KL-regularized RLHF objective and its closed-form optimal policy, DPO as implicit RL, PPO-for-LLMs anatomy, GRPO derivation, reward hacking simulation, and a unified view of the alignment landscape. |
