# Application Development

App development is where software engineering, systems architecture, and applied AI converge into something a user can actually touch. Most ML practitioners are trained on models and experiments but have little exposure to the plumbing that surrounds them — the APIs, databases, background workers, deployment pipelines, and real-time data flows that determine whether a product actually works at scale. This series closes that gap.

## About This Series

This series covers the [theory and practice]{.mark} of building **data-driven applications**: from the first HTTP endpoint through Docker-compose dev environments, async pipelines, vector search, and multimodal AI. It is organized around two threads that run in parallel: (1) general software engineering concepts applicable to any application, and (2) a concrete project — an AI-powered photo library — that progressively assembles those concepts into a working system.

**Audience:** Someone trained in machine learning or AI engineering, and so are comfortable with Python and NumPy, have worked with APIs or cloud services, and understand the ML stack. What you want is a grounded, practice-first treatment of the application layer: HTTP services, relational databases, event-driven data pipelines, containerization, and the integration patterns that hold them together.

**Stack.** The series uses tools chosen for clarity and real-world relevance: [Flet](https://flet.dev) for cross-platform desktop UI, [FastAPI](https://fastapi.tiangolo.com) for the API layer, [PostgreSQL](https://www.postgresql.org) + [pgvector](https://github.com/pgai/pgvector) for structured and vector storage (via Docker), [boto3](https://boto3.amazonaws.com/v1/documentation/api/latest/index.html) for S3 access, [CLIP](https://github.com/openai/CLIP) for local embeddings, and GPT-4o-mini for multimodal summaries. The goal is a minimal, composable stack where each layer is visible and teachable — not a framework that hides the complexity.

**The main project** is a [personal photo library application]{.mark} — a ground-up reimplementation of the ideas behind Apple Photos — that ingests 1 TB of photos from S3, extracts EXIF metadata and CLIP embeddings, clusters faces into people albums, and supports natural-language queries and AI-generated narrative summaries of time periods. The project is deliberately ambitious: the challenges it poses (batch ETL at scale, approximate nearest-neighbor search, async progress streaming, multimodal retrieval) are exactly the challenges that distinguish senior engineers from junior ones.

The goal is a system where submitting a training job to a RunPod pod, an AWS instance, or a bare SSH host looks and feels identical, with progress, logs, and outputs streaming back to a local Flet UI in real time.

## Prerequisites

- **Python 3.13+**, comfort with async/await, basic familiarity with HTTP
- **Cloud fundamentals:** S3 object storage, environment variables, basic IAM / credentials
- **Part I (01, 03)** is self-contained; Parts II–V assume you have read Part I or have equivalent Flet experience
- **Part III onward** assumes you have a working Docker installation and an AWS account with an S3 bucket
- No GPU is required: CLIP runs on CPU for the purposes of this series

## How to Read This Series

**If you are new to app development:** Start at [01](/notebooks/apps/01-flet.html) and work forward sequentially. Each notebook prepares the ground for the next. The Flet notebooks (Part I) are lighter warm-up material; the real depth begins at [05](/notebooks/apps/05-fastapi.html).

**If you already know Flet:** Skip 01 and 03 and start at [05](/notebooks/apps/05-fastapi.html). Refer back to the Flet notebooks only when the integration notebooks ([11](/notebooks/apps/11-integration.html)–[14](/notebooks/apps/14-photo-app.html)) introduce new UI patterns.

**If you only want the photo app:** Read [05](/notebooks/apps/05-fastapi.html) → [06](/notebooks/apps/06-docker.html) → [07](/notebooks/apps/07-database.html) → [08](/notebooks/apps/08-s3-pipeline.html) → [09](/notebooks/apps/09-embeddings.html) → [10](/notebooks/apps/10-clustering.html) → [11](/notebooks/apps/11-integration.html) → [12](/notebooks/apps/12-realtime.html) → [13](/notebooks/apps/13-multimodal.html) → [14](/notebooks/apps/14-photo-app.html) in order. Each notebook adds one layer to the same codebase.

**If you are here for the tooling:** [NBX](/notebooks/apps/nbx/index.html) is self-contained and can be read independently of the photo app project. For research infrastructure tooling, see the [Tooling](/notebooks/tooling/index.html) series.
