# NBX: Compute Platform

Most ML engineers eventually hit the same wall: a local machine is not enough, cloud consoles are too slow, and stitching together SSH commands, environment variables, and artifact downloads by hand does not scale. **NBX** is a compute platform designed to solve this — not a thin CLI wrapper around SSH, but a proper platform with a machine registry, a task queue, a live monitoring dashboard, and multi-provider dispatch. The goal is a system where submitting a training job to a RunPod pod, an AWS instance, or a bare SSH host looks and feels identical, and where progress, logs, and outputs flow back to a local Flet UI in real time.

This mini-series builds NBX from scratch. It assumes familiarity with FastAPI, Docker, and async Python from the main [App Development](/notebooks/apps/index.html) series. The system is deliberately production-grade: each design choice is motivated, each abstraction is justified, and the resulting codebase is something you could extend and deploy.
