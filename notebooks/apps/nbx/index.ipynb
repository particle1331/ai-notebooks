# NBX: Compute Platform

Most ML engineers eventually hit the same wall: a local machine is not enough, cloud consoles are too slow, and stitching together SSH commands, environment variables, and artifact downloads by hand does not scale. **NBX** is a [production-grade compute platform]{.mark} built to solve this — a system with a machine registry, a task queue, a live monitoring dashboard, and multi-provider dispatch where submitting a training job to a RunPod pod, an AWS instance, or a bare SSH host looks and feels identical, and where progress, logs, and outputs stream back to a local Flet UI in real time. This mini-series builds NBX from scratch.

## About This Series

**Goal.** To build a [real, deployable ML compute platform]{.mark} — not a toy — from the first design sketch through a working multi-provider system. Every abstraction is motivated, every design choice is justified, and the resulting codebase is something you could extend and put into production.

**Audience.** ML engineers and practitioners who have worked through the main [App Development](/notebooks/apps/index.html) series (or have equivalent experience with FastAPI, Docker, and async Python). You are comfortable building backend services and want to apply that skill to real research infrastructure — not another tutorial CRUD app, but a platform you would actually use.

**Stack.** [Flet](https://flet.dev) for the live resource-monitoring dashboard UI, [FastAPI](https://fastapi.tiangolo.com) for the REST and WebSocket layer, [asyncssh](https://asyncssh.readthedocs.io) for remote command execution, [papermill](https://papermill.readthedocs.io) for parameterized notebook runs, [pydantic-settings](https://docs.pydantic.dev/latest/concepts/pydantic_settings/) for config and secrets management, and [RunPod](https://www.runpod.io) / AWS / bare SSH as interchangeable provider targets.

## How to Read This Series

**If you want to understand the architecture first:** Read [01](/notebooks/apps/nbx/01-architecture.html) carefully before anything else. The provider abstraction and job spec schema established there underpin every subsequent notebook.

**If you are here for the Flet dashboard:** [02](/notebooks/apps/nbx/02-registry.html) is the most UI-heavy notebook. It stands alone if you already understand the `MachineRegistry` concept from 01.

**If you care about remote job execution:** [03](/notebooks/apps/nbx/03-packaging.html) and [04](/notebooks/apps/nbx/04-queue.html) are the core. Read them in order.

**If you want the full streaming story:** [05](/notebooks/apps/nbx/05-streaming.html) builds on all prior notebooks; read it last.
